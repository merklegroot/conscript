# Steam Deck build

Runs [`build_steam_deck.sh`](build_steam_deck.sh) to publish **Conscript** for `linux-x64` (self-contained single file).

Run the code cell below; output streams here as the build progresses.

```python
import subprocess
from pathlib import Path


def find_repo_root() -> Path:
    for directory in [Path.cwd(), *Path.cwd().parents]:
        if (directory / "Conscript" / "Conscript.csproj").is_file():
            return directory
    raise FileNotFoundError(
        "Could not find repo root (expected Conscript/Conscript.csproj). "
        "Open this notebook from the conscript repo."
    )


repo_root = find_repo_root()
script = repo_root / "scripts" / "build_steam_deck.sh"
if not script.is_file():
    raise FileNotFoundError(f"Missing build script: {script}")

print(f"Repo:   {repo_root}")
print(f"Script: {script}\n")

completed = subprocess.run(
    ["bash", str(script)],
    cwd=repo_root,
    check=False,
)

if completed.returncode != 0:
    raise RuntimeError(f"Build failed (exit code {completed.returncode})")

publish_dir = (
    repo_root
    / "Conscript"
    / "bin"
    / "Release"
    / "net10.0"
    / "linux-x64"
    / "publish"
)
print(f"\nPublish output: {publish_dir}")
if publish_dir.is_dir():
    for path in sorted(publish_dir.iterdir()):
        size_mb = path.stat().st_size / (1024 * 1024)
        print(f"  {path.name:<24} {size_mb:8.2f} MB")
else:
    print("  (directory not found)")
```
